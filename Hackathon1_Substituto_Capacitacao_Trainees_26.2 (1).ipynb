## O que construir

Criem uma **classe** chamada `GerenciadorCampeonato` que receba a lista `registros` na criação do objeto e implemente os seguintes métodos:

1. **`participantes_do_time(nome_time)`**: retorna uma lista com os nomes dos jogadores daquele time que participaram.

4. **`media_pontos(nome_jogo=None)`**: calcula a média de pontos. Se `nome_jogo` for informado, calcula só para aquele jogo.

**Atenção:** só quem **participou** entra nas contas. Quem não participou não conta em nenhum método.

Para guardar "o melhor até agora", comecem com `melhor_pontos = None` e vão atualizando dentro do loop quando encontrarem um valor maior.

````python
def calcular_media(valores):
    return sum(valores) / len(valores)

class GerenciadorCampeonato:
    def __init__(self, registros):
        self.registros = registros

    def _participantes(self, nome_jogo):
        return [
            r for r in self.registros
            if r["participou"] and r["pontos"] is not None 
            and (nome_jogo is None or r["jogo"] == nome_jogo)
        ]
        
    def participantes_do_time(self, nome_time):
        participantes = []

        for registro in self.registros:
            
            if registro["time"] == nome_time and registro["participou"] == True:
                participantes.append(registro["jogador"])

        return participantes

    def pontos_por_time(self):
        totais = {}

        for r in self.registros:
            if not r["participou"]:
                continue
            
            totais[r["time"]] = totais.get(r["time"], 0) + r["pontos"]
        
        return totais 

    def melhor_jogador(self, nome_jogo=None):
        candidatos = self._participantes(nome_jogo)
        melhor = ("Candidato", 0)

        for c in candidatos:
            if c["pontos"] > melhor[1]:
                melhor = (c["jogador"], c["pontos"])

        return melhor

    def media_pontos(self, nome_jogo=None):
        candidatos = self._participantes(nome_jogo)
        if not candidatos:
            return 0

        pontos = [c["pontos"] for c in candidatos]
        return calcular_media(pontos)
````
